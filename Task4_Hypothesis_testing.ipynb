```python
#Import the required libraries
import pandas as pd
from scipy import stats
```

```python
#Read your Excel dataset
df = pd.read_excel("Cleaned_Sales_Dataset (1).xlsx")

df.head()
```

```python
df['Gender'].value_counts()
```

Write your Hypothesis

Business Question:

Is there a significant difference in average Total Sales between Male and Female customers?

H₀ — Null Hypothesis:

There is no significant difference in average Total Sales between Male and Female customers.

H₁ — Alternative Hypothesis:

There is a significant difference in average Total Sales between Male and Female customers.

```python
#Separate Male and Female Sales

male_sales = df[df['Gender'] == 'Male']['Total_Sales']

female_sales = df[df['Gender'] == 'Female']['Total_Sales']
```

```python
#Perform the T-Test
t_statistic, p_value = stats.ttest_ind(
    male_sales,
    female_sales,
    equal_var=False
)

print("T-Statistic:", t_statistic)
print("P-Value:", p_value)
```

```python
#Get the Average Sales
df.groupby('Gender')['Total_Sales'].mean()
```
