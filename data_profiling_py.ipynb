```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Order_Date"].head()
```

```python
df["Month"] = df["Order_Date"].dt.month_name()

df[["Order_Date", "Month"]].head()
```

```python
import numpy as np
```

```python
df["Sales_Category"] = np.where(
    df["Total_Sales"] > df["Total_Sales"].median(),
    "High",
    "Low"
)

df[["Total_Sales", "Sales_Category"]].head()
```

```python
df["Sales_Category"].value_counts()
```

```python
df.to_excel("Cleaned_Sales_Dataset.xlsx", index=False)

print("File saved successfully")
```
